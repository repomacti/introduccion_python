La información que vemos cotidianamente en internet, redes sociales, móviles y computadoras personales, está organizada en diferentes tipos y formatos como en hojas de cálculo, imágenes, audio, video, entre otros. La mayor parte de toda esta información está compuesta de texto y números.

## Tipos básicos y variables. {#sec-l1-tipos-basicos-variables}

::: {#exm-l1-ej1 .callout-tip icon=false}
## {{< iconify vscode-icons:file-type-light-mypy >}} Índice de masa corporal. 

El índice de masa corporal (IMC) asocia la masa (peso) con la talla de una persona. La fórmula para medir el IMC nació gracias al matemático, astrónomo y estadístico belga <a href="https://en.wikipedia.org/wiki/Adolphe_Quetelet"  target="_blank" rel="noopener">Lambert Adolphe Quetelet</a> en 1832 y se define como sigue:

$$
\text{IMC} = \dfrac{\text{peso [kg]}}{(\text{estatura [m])}^2}
$$

En este ejemplo vamos a calcular el IMC dados el peso y la estatura de una persona.
:::

En Python es posible definir una variable (un nombre) para almacenar un valor. Por ejemplo, definamos una variable para almacenar el peso de una persona:

````python
peso = 85.0
````

Para imprimir el contenido de una variable podemos usar la función  `print()` como sigue:

````python
print(peso)
````

También podemos saber de qué tipo es la variable `peso` con la función `type()`:

````python
type(peso)
````

Observa que la variable `peso`es de tipo `float`, que es uno de los tipos de datos básicos en Python.

::: {#tip-l1-print_type .callout-tip}

Como puedes ver, al ejecutar la celda con el código `type(peso)` se imprime el tipo de la variable. Esto funciona de esta manera porque estamos en un ambiente interactivo (Jupyter Notebook), sin embargo conviene siempre agregar la función `print()` para que la impresión del resultado funcione en cualquier tipo de ambiente. Es decir, el código más funcional sería: `print(type(peso))`. En lo que sigue usaremos `print()` para imprimir el resultado de una declaración.
:::

Definamos ahora una variable para almacenar la estatura de una persona, imprimamos el contenido de la variable y su tipo:

````python
# Escribimos las tres instrucciones en una sola celda:
estatura = 1.88
print(estatura)
print(type(estatura))
````

Algunas observaciones del código anterior:

* La línea que aparece con un `#` al inicio es un comentario, es decir, el intérprete de Python no la ejecuta por lo que es  útil solamente para documentar las acciones del código.
* Se imprime `1.88` que es el contenido de la variable `peso` y corresponde al código `print(estatura)`.
* Se imprime `<class 'float'>` que es el resultado del código `print(type(estatura))`.
* Nota que usando `print` antes de `type` obtenemos un poco más de información que solo usando `type`. En esta última declaración estamos concatenando dos funciones, esto lo explicamos con detalle en la [sección @sec-l1-composicion-de-funciones].

La función `print()` puede recibir más de un argumento, por ejemplo:

````python
print("Peso =", peso)
print("Estatura =", estatura)
````

Observa que el primer argumento en ambas líneas son textos los cuales se definen usando comillas `"..."`. El segundo argumento son las variables antes definidas `peso` y `estatura`. Podemos proporcionar más argumentos a la función `print()` por ejemplo: 

````python
# Ahora agregamos las unidades
print("Peso =", peso, "[kg]")
print("Estatura =", estatura, "[m]")
````

El resultado del código anterior da más información de las variables.

Recordemos que deseamos calcular el IMC cuya fórmula es:

$$
\text{IMC} = \dfrac{\text{peso [kg]}}{(\text{estatura [m])}^2}
$$

En Python esta operación se realiza como sigue:

````python
# Calcular el IMC
IMC = peso / estatura**2
````

Hemos usado dos operadores: `/` para división y `**` para calcular una potencia. Hablaremos más de estos operadores en la [sección @sec-l1-operadores-aritmeticos].

Para saber cuál fue el resultado de nuestro cálculo hagamos lo siguiente:

````python
print("IMC =", IMC)
print(type(IMC))
````

Observa en la celda anterior que en cada línea de código agregamos una cadena, la variable y el resultado de la función `type()` como argumentos de la función `print()`. En cada caso el tipo de variable es distinto: `int`, `str` y `bool`. Estos tres son tipos básicos en Python. 

En la siguiente celda obtenemos el **contenido** de todas las variables definidas hasta el momento:

En la siguiente celda obtenemos el **tipo** de todas las variables definidas hasta el momento:

## Operadores aritméticos. {#sec-l1-operadores-aritmeticos}

En el [ejemplo @exm-l1-ej1] usamos los operadores: `/` y `**` para calcular el IMC. Para los tipos numéricos básicos que acabamos de revisar, existen más operaciones aritméticas que se pueden aplicar sobre ellos. Estas operaciones están organizadas como se describe en la [sección @sec-a-operadores-aritmeticos].  A continuación veremos un ejemplo más interesante del uso de estos operadores. 

::: {#exm-ej1 .callout-tip icon=false}
## {{< iconify vscode-icons:file-type-light-mypy >}} Cálculo del interés compuesto.

El **interés compuesto** es aquel que se va sumando al capital inicial y sobre el que se van generando nuevos intereses. El dinero, en este caso, tiene un efecto multiplicador porque los intereses producen nuevos intereses. La fórmula para calcular el interés compuesto es la siguiente:

$$
A = P (1 + \dfrac{r}{n})^{nt}
$$

donde:

* $A$ = cantidad final
* $P$ = capital inicial
* $r$ = tasa de interés
* $n$ = número de veces que se ha aplicado el interés por periodo
* $t$ = número de periodos transcurridos

Vamos a calcular el interés compuesto con los siguientes datos:

* $t = 2$,
* $n = 12$,
* $P = 250,000$ y
* $r = 0.18$.

:::

Lo primero para realizar este cálculo es definir las variables:

````python
t = 2
n = 12
P = 250_000 # formato de entero grande
r = 0.18
````

Verificamos el contenido y el tipo de cada variable:

````python
print(t, type(t), id(t))
print(n, type(n), id(n))
print(P, type(P), id(P))
print(r, type(r), id(r))
````

Ahora implementamos la fórmula del interés compuesto: $A = P (1 + \dfrac{r}{n})^{nt}$

````python
A = P * (1 + (r/n)) ** (n * t)
````

````python
print("Cantidad final A = ", A)
````

::: {#tip-l1-precedencia .callout-tip}
Observa el uso de paréntesis en la implementación de la fórmula. Este uso de paréntesis es importante para generar el resultado correcto. Algunas veces se pueden omitir los paréntesis, pero se debe tener claro el nivel de precedencia de los operadores, revisa la [sección @sec-a-precedencia] para más información al respecto.
:::

## Objetos y tipado dinámico. {#sec-l1-objetos-tipado-dinamico}

Python es un lenguaje *totalmente* Orientado a Objetos lo que significa que todo lo que definimos será un objeto de una clase determinada. Python tiene definidas clases de las cuales es posible construir objetos. 

Por ejemplo, la variable `peso` definida en la [sección @sec-l1-tipos-basicos-variables] representa el nombre de un objeto de tipo `float`:

````python
print(peso, type(peso), id(peso))
````

Observa que el nombre del objeto es `peso`, el tipo es de la clase `<class 'float'>`, su contenido es `85.0` y su identificador en memoria es `4386712720`. El nombre identifica al objeto dentro del código, el identificador lo distigue dentro de la memoria.

Ahora hagamos lo siguiente:

````python
masa = peso
print(peso, type(peso), id(peso))
print(masa, type(masa), id(masa))
````

Observa que los nombres `peso` y `masa` hace referencia al mismo objeto pues tienen el mismo identificador. 

Los operadores `is` y `not` también pueden ayudar a saber cuando dos objetos son el mismo:

````python
peso is masa
````

````python
peso is not masa
````

````python
# ¿Qué sucederá ahora?
masa = 5
print(peso, type(peso), id(peso))
print(masa, type(masa), id(masa))
````

Ahora los objetos son diferentes, también los tipos de `peso` y `masa` son diferentes. 

Podemos usar la función incorporada `isinstance()` para comprobar de qué tipo son las variables:

````python
print(isinstance(peso, float))
print(isinstance(masa, float))
````

````python
print(isinstance(peso, int))
print(isinstance(masa, int))
````

````python
# ¿y ahora?
masa = 3 + 5j

print(isinstance(peso, complex))
print(isinstance(masa, complex))
````

````python
# Vamos más allá
masa = "En realidad no es lo mismo que peso"
print(peso, type(peso), id(peso))
print(masa, type(masa), id(masa))
print(isinstance(peso, float))
print(isinstance(masa, str))
````

::: {#tip-l1-tipado-dinamico .callout-tip}

Como se puede observar el tipo de `masa` cambia dependiendo de su contenido, esto es lo que se conoce como **tipado dinámico**.
:::

### Atributos y métodos. {#sec-l1-atributos-metodos}

Una clase define atributos y métodos para los objetos.

Por ejemplo, definimos tres objetos:

````python
entero = 1
real = 1.1
complejo = 1 + 2j
````

Cada objeto tiene un conjunto de atributos y métodos los cuales dependen del tipo del objeto. Por ejemplo los objetos de tipo entero tienen el método `as_integer_ratio()` que se aplica como sigue:

````python
entero.as_integer_ratio()
````

El resultado de este método es una tupla que contiene dos números que la ser divididos generan el número original.

El objeto de tipo flotante también tiene un método similar:

````python
real.as_integer_ratio()
````

Podemos comprobar que el resultado es correcto como sigue:

````python
2476979795053773 / 2251799813685248
````

Obtenemos el número real original.

Para los número complejos también tenenmos varios métodos:

````python
complejo.conjugate()
````

Y también se tienen atributos:

````python
complejo.real # la parte real
````

````python
complejo.imag # la parte imaginaria
````

### Interacción entre objetos.

En este estilo de programación orientado a objetos, es posible tener una intereacción entre objetos que es similar a la explicada en la composición de función [sección @sec-l1-composicion-de-funciones]. 

Por ejemplo, aplicamos el método `as_integer_ratio()` al objeto `real` y su resultado se almacena en `ratio`:  

````python
ratio = real.as_integer_ratio()
print(ratio, type(ratio))
````

Observamos que el resultado es un objeto de tipo `tuple` el cual tiene también métodos, por ejemplo `count(a)` cuenta cuántos elementos iguales al argumento `a` existen en la tupla, en este caso podemos probar lo siguiente:

````python
ratio.count(2476979795053773)
````

Podemos realizar las operaciones anteriores en una sola línea como sigue:

````python
real.as_integer_ratio().count(2476979795053773)
````

El estilo de programación anterior se explica en la [figura @fig-l1-metodos_objetos_python].
