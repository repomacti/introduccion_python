## Ejercicio 1.


Cuando un objeto es lanzado en tiro vertical hacia arriba a una velocidad inicial $v_0$, su posición y velocidad después de un tiempo $t$ se pueden calcular con las siguientes dos fórmulas:

$$
\begin{eqnarray}
\text{Posición: } \; y & = & v_0 t - \dfrac{g t^2}{2} \tag{1}\\
\text{Velocidad: } \; v_f & = & v_0 - g t \tag{2}
\end{eqnarray}
$$

donde $t$ es el tiempo que ha pasado desde el lanzamiento inicial, $g = 9.81 \text{ [m/s}^2\text{]}$ es la aceleración de la gravedad y $v_f$ la velocidad del objeto al tiempo $t$.

<img src = "./tiro_vertical.png" width=400px />

1. Escribir la función `tiroVertical(t, v0)`:
    * Debe recibir como parámetros el tiempo `t` y la velocidad inicial `v0`.
    * Debe regresar como resultado la posición $y$ y velocidad $v_f$ en el tiempo `t`, calculadas con las fórmulas $(1)$ y $(2)$.
    * Posteriormente evaluar la fórmula para $v_0 = 20 \text{ [m/s]}$ y $t = 1.0 \text{ [s]}$.

</div>

````python
# def tiroVertical(t, v0):
# ...

### BEGIN SOLUTION
from macti.eval import FileAnswer
file_answer = FileAnswer() 

def tiroVertical(t, v0):       # Recibe como parámetros el tiempo y la vel. inicial
    g = 9.81                    # Aceleración de la gravedad [m/s^2]
    y = v0 * t - 0.5 * g * t**2 # Posición
    vf = v0 - g * t             # Velocidad
    return (y, vf)              # Regresa la posición [m] y la velocidad [m/s]
### END SOLUTION
````

````python
# Evaluamos la función

### BEGIN SOLUTION
t = 1.0    # [s]
v0 = 20.0  # [m/s]

y, vf = tiroVertical(t, v0)

file_answer.write('1', y, "La posición no está correctamente calculada, checa la fórmula")
file_answer.write('2', vf, "La velocidad no está correctamente calculada, checa la fórmula")
file_answer.to_file('q4')
### END SOLUTION

print(f"Posición (t = 1.0, v0 = 20)= {y}")
print(f"Velocidad (t = 1.0, v0 = 20)= {vf}")
````

5. Calcular la posición y la velocidad en los tiempos $\{t| t = t_{ymax}+i*\delta\}$, con $i$ un número entero en el rango $[-10, 10]$ y $\delta = 0.01$. Almacenar los tiempos, las posiciones y las velocidades así calculadas en las listas: `t_l`, `y_l` y `v_l`.

</div>

````python
# 1. Genera la lista de tiempos.

### BEGIN SOLUTION
delta = 0.01
t_l = []

for i in range(-10,11):
    t_l.append(t_ymax + i*delta)

# Forma Pythónica
#[t_l.append(t_ymax + i*delta) for i in range(-10, 11)]

file_answer.write('3', t_l, "La lista de los tiempos no corresponde con el conjunto descrito en el ejercicio")
file_answer.to_file('q4')
### END SOLUTION

print(f"Lista de tiempos ({len(t_l)}):\n", t_l, )
````

````python
# 2. Calcula las posiciones y velocidades
y_l = []
v_l = []

### BEGIN SOLUTION
for i, v in enumerate(range(-10,11)):
    pos, vel = tiroVertical(t_l[i], v0)
    y_l.append(pos)
    v_l.append(abs(vel))

# Forma más simple
#for t in t_l:
#    pos, vel = tiroVertical(t, v0)
#    y_l.append(pos)
#    v_l.append(abs(vel))

file_answer.write('4', y_l, "La lista de las posiciones no está correctamente calculada")
file_answer.write('5', v_l, "La lista de las velocidades no está correctamente calculada")
file_answer.to_file('q4')
### END SOLUTION

print("tiempos : \n{}".format(t_l))
print("\nposiciones: \n{}".format(y_l))
print("\nvelocidades: \n{}".format(v_l))
````
